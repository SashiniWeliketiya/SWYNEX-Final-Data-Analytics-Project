```python
import pandas as pd
import numpy as np
df = pd.read_csv('Mall_Customers.csv')

print("Initial data shape:", df.shape)

df.loc[0, 'Annual_Income_(k$)'] = np.nan

df = pd.concat([df, df.iloc[[1]]], ignore_index=True)

print("Shape after introducing flaws (Missing & Duplicates):", df.shape)
```

```python
df['Annual_Income_(k$)'] = df['Annual_Income_(k$)'].fillna(df['Annual_Income_(k$)'].median())
```

```python
df = df.drop_duplicates()
```

```python
df['Genre'] = df['Genre'].str.strip().str.title()
```

```python
df['CustomerID'] = df['CustomerID'].astype(int)
df['Age'] = df['Age'].astype(int)
```

```python
df.to_csv('cleaned_mall_customers.csv', index=False)

print("Data cleaning completed successfully. Cleaned shape:", df.shape)
```
